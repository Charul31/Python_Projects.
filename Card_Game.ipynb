```python
import random

suits = ('Hearts', 'Diamonds', 'Spades', 'Clubs')
ranks = ('Two', 'Three', 'Four', 'Five', 'Six', 'Seven', 'Eight', 'Nine', 'Ten', 'Jack', 'Queen', 'King', 'Ace')

values = {'Two':2 , 'Three':3, 'Four':4, 'Five':5, 'Six':6, 'Seven':7, 'Eight':8, 'Nine':9, 'Ten':10, 'Jack':11, 'Queen':12, 'King':13, 'Ace':14}
```

```python
class Card:
    def __init__(self,suit,rank):
        self.suit=suit
        self.rank=rank
        self.value= values[rank]

    def __str__(self):
        return self.rank + ' of ' + self.suit
```

```python
class Deck():
    def __init__(self):
        self.all_cards=[]

        for suit in suits:
            for rank in ranks:
                #create card object
                created_card= Card(suit,rank)
                self.all_cards.append(created_card)  # till here every card will be in order so now we've to shuffle the deck

    def shuffle(self):
        random.shuffle(self.all_cards)

    # Now we want to pick one card

    def deal_one(self):
        return self.all_cards.pop()
```

```python
class Player():
    def __init__(self, name):
        self.name=name
        self.all_cards=[]

    def remove_card(self):
        return self.all_cards.pop(0)

    def add_card(self, new_cards):
        
        if type(new_cards)== type([]):      #To add a list of cards
            self.all_cards.extend(new_cards)

        else:
            self.all_cards.append(new_cards)    #To add single card
        

    def __str__(self):
        return f'Player {self.name} has {len(self.all_cards)} cards'
```
