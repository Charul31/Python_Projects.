```python
class Account():
    def __init__(self, owner,balance=0):
        self.owner=owner
        self.balance=balance

    def __str__(self):
        return (f'Account owner:    {self.owner} \nAccount balance:  ${self.balance}')

    def deposit(self,add):
        self.balance += add
        self.add =add
        print(f'Deposit Accepted Rs.{self.add}')

    def withdraw(self, wd):
        self.wd=wd
        if wd<=self.balance:            
            self.balance -= wd
            print (f'Money withdrawn  Rs.{self.wd}')
        else:
            print('Funds Unavailable')
```

```python
acct1 = Account('Charul',100)
```
